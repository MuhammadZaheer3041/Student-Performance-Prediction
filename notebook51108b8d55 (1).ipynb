```python
shifted_rows = df[df["rating"].isin(["74 min", "84 min", "66 min"])]
shifted_rows
```

```python
shift_mask = df["rating"].isin(["74 min","84 min", "66 min"])
```

```python
df.loc[shift_mask, "duration"] = df.loc[shift_mask, "rating"]
df.loc[shift_mask, "duration"] = None
```

```python
df.loc[shift_mask]
```

```python
df["rating"].value_counts(dropna = False)
```

```python
df.groupby("type")["rating"].value_counts()
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
df["rating"].isna().sum()
```

```python
df.isna().sum()
```

```python
for col in ["director","cast","country"]:
    df[col] = df[col].fillna("unkown")
```

```python
df[["director","cast","country"]].isna().sum()
```

```python
(df["director"] == "Unknown").sum(), ((df["director"]=="Unknown") & (df["cast"]=="Unknown")).sum()
```

```python
df["date_added"]
```

```python
df["date_added"]= df["date_added"].str.strip()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"],format = "%B %d, %Y")
```

```python
df["date_added"].head()
```

```python
df["date_added"].isna().sum()
```

```python
df[df["date_added"].isna()]
```

```python
df[df["date_added"].isna()][["title","type","release_year"]]
```

```python
df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month
```
